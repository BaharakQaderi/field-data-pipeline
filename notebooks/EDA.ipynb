### Timeseries Analysis for July 16 Flight Data (11:47:46 - 12:18:40)
Creating a timeseries visualization for the FLIGHT_SEGMENT_loadcells_force field during the specified time range.

```python
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.dates as mdates
from datetime import datetime, timedelta
import numpy as np

# Load the merged data for July 16
# If already loaded as merged_df_16, we can use that, otherwise load it again
if 'merged_df_16' not in globals():
    try:
        merged_df_16 = pd.read_csv('../outputs/merged_flight_data_2025-07-16.csv')
    except Exception as e:
        print(f"Error loading file: {e}")

# Check if data is available
if 'merged_df_16' in globals():
    # Convert timestamp to datetime if it's not already
    if 'timestamp' in merged_df_16.columns:
        timestamp_col = 'timestamp'
    else:
        # If no explicit timestamp column, look for a column that might contain time information
        potential_time_cols = [col for col in merged_df_16.columns if 'time' in col.lower()]
        if potential_time_cols:
            timestamp_col = potential_time_cols[0]
            print(f"Using {timestamp_col} as timestamp column")
        else:
            print("No timestamp column found. Creating one based on index.")
            # Create a timestamp column based on the index, assuming a regular time interval
            merged_df_16['timestamp'] = pd.date_range(start='2025-07-16 00:00:00', periods=len(merged_df_16), freq='1s')
            timestamp_col = 'timestamp'
    
    # Ensure timestamp is datetime type
    if pd.api.types.is_string_dtype(merged_df_16[timestamp_col]):
        # Use format='mixed' to handle different timestamp formats
        merged_df_16[timestamp_col] = pd.to_datetime(merged_df_16[timestamp_col], format='mixed')
    
    # Specify the time range from the flight metrics
    start_time_str = '2025-07-16 11:47:46'
    end_time_str = '2025-07-16 12:18:40'
    
    # Convert to datetime
    start_time = pd.to_datetime(start_time_str)
    end_time = pd.to_datetime(end_time_str)
    
    # Filter data for the specified time range
    filtered_df = merged_df_16[(merged_df_16[timestamp_col] >= start_time) & 
                              (merged_df_16[timestamp_col] <= end_time)]
    
    if len(filtered_df) > 0:
        # Create the timeseries plot
        plt.figure(figsize=(14, 6))
        plt.plot(filtered_df[timestamp_col], filtered_df['FLIGHT_SEGMENT_loadcells_force'], 
                 linewidth=1.5, color='blue')
        
        # Set plot title and labels
        plt.title(f'Timeseries of FLIGHT_SEGMENT_loadcells_force (July 16, {start_time.strftime("%H:%M:%S")} - {end_time.strftime("%H:%M:%S")})', 
                 fontsize=14)
        plt.xlabel('Time', fontsize=12)
        plt.ylabel('Force', fontsize=12)
        
        # Format x-axis to show time nicely
        plt.gca().xaxis.set_major_formatter(mdates.DateFormatter('%H:%M:%S'))
        plt.xticks(rotation=45)
        
        # Add grid for better readability
        plt.grid(True, alpha=0.3)
        
        # Display statistics
        avg_force = filtered_df['FLIGHT_SEGMENT_loadcells_force'].mean()
        max_force = filtered_df['FLIGHT_SEGMENT_loadcells_force'].max()
        min_force = filtered_df['FLIGHT_SEGMENT_loadcells_force'].min()
        
        stats_text = f"Statistics:\nAvg Force: {avg_force:.2f}\nMax Force: {max_force:.2f}\nMin Force: {min_force:.2f}\nData Points: {len(filtered_df)}"
        plt.figtext(0.02, 0.02, stats_text, fontsize=10, bbox=dict(facecolor='white', alpha=0.8))
        
        # Tight layout to use space efficiently
        plt.tight_layout()
        plt.show()
        
        print(f"Plotted {len(filtered_df)} data points from {start_time} to {end_time}")
    else:
        print(f"No data found for the specified time range: {start_time} - {end_time}")
        print(f"Available time range: {merged_df_16[timestamp_col].min()} - {merged_df_16[timestamp_col].max()}")
        
        # Let's check if there's any data for July 16, 11:47
        date_str = '2025-07-16'
        if any(merged_df_16[timestamp_col].dt.strftime('%Y-%m-%d') == date_str):
            print(f"Data is available for {date_str}, but not for the specific time range.")
            # Show available hours for that day
            available_hours = merged_df_16[merged_df_16[timestamp_col].dt.strftime('%Y-%m-%d') == date_str][timestamp_col].dt.hour.unique()
            print(f"Available hours for {date_str}: {sorted(available_hours)}")
else:
    print("merged_df_16 not available. Please load the dataset first.")
```

### Timeseries Analysis for Force Data (18:50 - 19:21)
Creating a timeseries visualization for the total_force_kg field from the loadcell readings on July 15, 2025.

```python
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.dates as mdates
from datetime import datetime
import pytz

# Load the loadcell readings data
file_path = '/Users/baharakqaderi/field-data-pipeline/data/FORCES/loadcell_readings_20250715_183235.csv'
df_forces = pd.read_csv(file_path)

# Display basic info about the dataframe
print(f"Dataframe shape: {df_forces.shape}")
print(f"Columns: {df_forces.columns.tolist()}")
print(f"First few rows:")
display(df_forces.head())

# Use the 'datetime' column as our timestamp column since it's already in datetime format
if 'datetime' in df_forces.columns:
    timestamp_col = 'datetime'
    print(f"Using {timestamp_col} column for time information")
elif 'timestamp' in df_forces.columns:
    timestamp_col = 'timestamp'
    print(f"Using {timestamp_col} column for time information")
else:
    # Look for any column that might contain timestamp information
    time_cols = [col for col in df_forces.columns if any(time_term in col.lower() for time_term in ['time', 'date'])]
    if time_cols:
        timestamp_col = time_cols[0]
        print(f"Using {timestamp_col} as the timestamp column")
    else:
        print("No timestamp column found. Creating a timestamp column based on file name.")
        # Extract datetime from filename: loadcell_readings_20250715_183235.csv
        file_date = "2025-07-15"
        file_start_time = "18:32:35"
        
        # Create a timestamp column assuming regular intervals
        df_forces['datetime'] = pd.date_range(
            start=f"{file_date} {file_start_time}", 
            periods=len(df_forces), 
            freq='10ms'  # Adjust the frequency based on your data sampling rate
        )
        timestamp_col = 'datetime'
        print(f"Created datetime column starting at {file_date} {file_start_time}")

# Convert timestamp to datetime format if it's not already
df_forces[timestamp_col] = pd.to_datetime(df_forces[timestamp_col])
    
# Define the specific time range
start_time_str = '2025-07-15 18:50:00'
end_time_str = '2025-07-15 19:21:00'

# Convert to datetime
start_time = pd.to_datetime(start_time_str)
end_time = pd.to_datetime(end_time_str)

print(f"Data timespan: {df_forces[timestamp_col].min()} to {df_forces[timestamp_col].max()}")
    
# Filter data for the specified time range
filtered_forces = df_forces[(df_forces[timestamp_col] >= start_time) & 
                          (df_forces[timestamp_col] <= end_time)]

if len(filtered_forces) > 0:
    print(f"Filtered data contains {len(filtered_forces)} rows")
    
    # Create the timeseries plot
    plt.figure(figsize=(14, 6))
    plt.plot(filtered_forces[timestamp_col], filtered_forces['total_force_kg'], 
            linewidth=1.5, color='green')
    
    # Set plot title and labels
    plt.title(f'Timeseries of total_force_kg (July 15, {start_time.strftime("%H:%M:%S")} - {end_time.strftime("%H:%M:%S")})', 
            fontsize=14)
    plt.xlabel('Time', fontsize=12)
    plt.ylabel('Force (kg)', fontsize=12)
    
    # Format x-axis to show time nicely
    plt.gca().xaxis.set_major_formatter(mdates.DateFormatter('%H:%M:%S'))
    plt.xticks(rotation=45)
    
    # Add grid for better readability
    plt.grid(True, alpha=0.3)
    
    # Calculate and display statistics
    avg_force = filtered_forces['total_force_kg'].mean()
    max_force = filtered_forces['total_force_kg'].max()
    min_force = filtered_forces['total_force_kg'].min()
    std_force = filtered_forces['total_force_kg'].std()
    
    stats_text = (f"Statistics:\nAvg Force: {avg_force:.2f} kg\n"
                 f"Max Force: {max_force:.2f} kg\n"
                 f"Min Force: {min_force:.2f} kg\n"
                 f"Std Dev: {std_force:.2f} kg\n"
                 f"Data Points: {len(filtered_forces)}")
    
    plt.figtext(0.02, 0.02, stats_text, fontsize=10, bbox=dict(facecolor='white', alpha=0.8))
    
    # Tight layout to use space efficiently
    plt.tight_layout()
    plt.show()
    
    # Print summary statistics
    print("\nSummary Statistics for total_force_kg:")
    print(filtered_forces['total_force_kg'].describe())
else:
    print(f"No data found for the specified time range: {start_time} - {end_time}")
    print("Available time range: ", df_forces[timestamp_col].min(), " to ", df_forces[timestamp_col].max())
    
    # Show a sample of timestamps to help debug
    print("\nSample timestamps from the dataframe:")
    print(df_forces[timestamp_col].head().tolist())
```
